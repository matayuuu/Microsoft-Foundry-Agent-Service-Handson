# Azure ML setup for Labs 7 / 8

Lab 1 で download した bundle を Azure ML Studio の **User files** へ upload した後に実行します。最初は built-in **Python 3.10 - SDK v2** kernel を選択してください。

```python
import json
import subprocess
import sys
from pathlib import Path


def find_bundle_root() -> Path:
    for candidate in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
        if (candidate / "scripts" / "setup_azureml.py").is_file():
            return candidate
    message = "Bundle root が見つかりません。"
    raise RuntimeError(f"{message} 展開した最上位 folder を User files へ upload してください。")


BUNDLE_ROOT = find_bundle_root()
print(BUNDLE_ROOT)
```

```python
context_path = BUNDLE_ROOT / ".workshop" / "context.json"
context = json.loads(context_path.read_text(encoding="utf-8"))
if context.get("setup_status") != "complete":
    raise RuntimeError("Lab 1 の setup が complete ではありません。")
if not isinstance(context.get("resource_outputs"), dict):
    raise RuntimeError("context.json に resource_outputs がありません。")
print("Cloud Shell provisioning context: complete")
```

```python
subprocess.run(
    [sys.executable, str(BUNDLE_ROOT / "scripts" / "setup_azureml.py")],
    cwd=BUNDLE_ROOT,
    check=True,
)
```
